#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)
tidy_data <- data %>% #keeps only the columns needed for the analysis
select(year, school, seed, conf_wins_pct, reg_wins_pct, tourney_wins) %>% 
#convert the percentage columns and the outcome to numeric.
#parse_number() pulls the number out of the text
mutate(
    conf_wins_pct = parse_number(as.character(conf_wins_pct)),
    reg_wins_pct = parse_number(as.character(reg_wins_pct)),
    tourney_wins = as.numeric(tourney_wins)) %>%
#convert seed to numeric, then create a categorical seed_group variable.
mutate(
    seed = as.numeric(seed),
    seed_group = case_when(
        seed <= 3 ~ "Top (1-4)",
        seed <= 8 ~ "Upper-mid (5-8)",
        seed <= 12 ~ "Lower-mid (9-12)",
        TRUE ~ "Bottom (13-16)"),
#turn seed_group into a factor with an explicit order so that plots and tables list the groups from best to worst seed
    seed_group = factor(seed_group, levels = c("Top (1-4)", "Upper-mid (5-8)", "Lower-mid (9-12)", "Bottom (13-16)"))) %>%
#remove rows with missing values in the key variables
drop_na(seed, conf_wins_pct, reg_wins_pct, tourney_wins) %>%
#reorder columns so the outcome and predictors are easy to find
relocate(year, school, tourney_wins, seed, seed_group, conf_wins_pct, reg_wins_pct)
#previews the first 6 rows of data
head(tidy_data, 6)
```

#### Task 5.2 Perform the Analysis (20 points)

```r
library(tidyverse)
library(tidymodels)
```

```r
#1. Tidy the data - Keep only the outcome and predictors, and drop rows with missing values
model_data <- tidy_data%>%
select(tourney_wins, seed, conf_wins_pct, reg_wins_pct) %>%
drop_na()
#2. Train/test split (75/25), stratified by the outcome
data_split <- initial_split(model_data,prop = 0.75, strata = tourney_wins)
train_data <- training(data_split)
test_data <- testing(data_split)
#3. Cross-validation folds on the training set only
folds <- vfold_cv(train_data, v = 10, strata = tourney_wins)
#4. Define the models
#multiple linear regression
lm_spec <- linear_reg() %>%
set_engine("lm")
#Baseline: always predicts the mean of tourney_wins
baseline_spec <- null_model() %>%
set_engine("parsnip") %>%
set_mode("regression")
#Workflows bundle the formula and the model
lm_wf <- workflow() %>%
add_formula(tourney_wins ~ seed + conf_wins_pct + reg_wins_pct) %>%
add_model(lm_spec)
baseline_wf <- workflow() %>%
add_formula(tourney_wins ~ seed + conf_wins_pct + reg_wins_pct) %>%
add_model(baseline_spec)
#5. Performance metrics: RMSE and R-sqared
reg_metrics <- metric_set(rmse, rsq)
#6. Estimate performance with cross-validation
cv_lm <- fit_resamples(lm_wf,resamples = folds, metrics = reg_metrics)
cv_base <- fit_resamples(baseline_wf,resamples = folds, metrics = reg_metrics)
#Compare average cv performance
cv_results <- bind_rows(
    collect_metrics(cv_lm) %>% mutate(model = "Linear regression"),
    collect_metrics(cv_base) %>% mutate(model = "Baseline (mean)")) %>%
select(model, .metric, mean, std_err)
print(cv_results)
#7. Final evaluation on the test set
final_lm <- last_fit(lm_wf, data_split, metrics = reg_metrics)
final_base <- last_fit(baseline_wf, data_split, metrics = reg_metrics)
test_results <- bind_rows(
    collect_metrics(final_lm) %>% mutate(model = "Linear regression"),
    collect_metrics(final_base) %>% mutate(model = "Baseline (mean)")) %>%
select(model, .metric, .estimate)
print(test_results)
#8. Coefficients of the final model
#Each coefficient = change in expected wins per 1-unit increase in that predictor
final_fit <- extract_workflow(final_lm)
coef_table <- tidy(final_fit, conf.int = TRUE)
print(coef_table)
#9. Check seed vs. win percentages)
cor(train_data %>% select(seed, conf_wins_pct, reg_wins_pct))
```
